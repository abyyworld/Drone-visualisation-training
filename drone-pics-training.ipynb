```python
import threading
import time
import subprocess

def keep_alive():
    while True:
        time.sleep(300)  # every 5 minutes
        try:
            # touches a file to signal activity
            subprocess.run(['touch', '/kaggle/working/keepalive.txt'])
            print(f"Still alive: {time.strftime('%H:%M:%S')}")
        except:
            pass

thread = threading.Thread(target=keep_alive, daemon=True)
thread.start()
print("Keep-alive thread started")
```
